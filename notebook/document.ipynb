```python
## create a simple txt file

import os
os.makedirs("../data/text_files", exist_ok = True)
```

```python
sample_texts = {
    "../data/text_files/python_intro.txt" : """ Python programming introduction
    Python is a high-level, general-purpose programming language that is widely celebrated for its clean, 
    English-like syntax and emphasis on code readability. 
    
    Created by Guido van Rossum and first released in 1991, 
    Python has evolved into one of the most dominant and heavily utilized languages globally. 
    It operates as an interpreted, dynamically-typed language managed under a non-profit 
    organization called the Python Software Foundation (PSF). 
    
    [1] (https://www.w3schools.com/python/python_intro.asp), 
    [2] (https://pythoninstitute.org/about-python), 
    [3] (https://www.coursera.org/in/articles/what-is-python-used-for-a-beginners-guide-to-using-python), 
    [4] (https://www.youtube.com/watch?v=mlTd1EPITBo&t=297), 
    [5] (https://en.wikipedia.org/wiki/Python_(programming_language)), 
    [6] (https://docs.python.org/3/faq/general.html)

      """
}

for filepath,content in sample_texts.items():
    with open(filepath, 'w', encoding="utf-8") as f:
        f.write(content)

print(" sample text files created")
```

```python
### TextLoader

#from langchain.document_loaders import TextLoader

from langchain_community.document_loaders import TextLoader

loader=TextLoader("../data/text_files/python_intro.txt", encoding="utf-8")
document=loader.load()
print(document)
```

```python
### Directory Loader

from langchain_community.document_loaders import DirectoryLoader
from langchain_community.document_loaders import TextLoader
from tqdm import tqdm

## load all text files from the directory

dir_loader=DirectoryLoader(
    "../data/text_files",
    glob="**/*.txt", # patterna to match files
    loader_cls= TextLoader, # loader class to use
    loader_kwargs={'encoding': 'utf-8'},
    show_progress=True,
    use_multithreading=True
)
documents=dir_loader.load()
documents
```
